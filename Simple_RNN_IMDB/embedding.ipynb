```python
#implementing word embedding with Keras Tensorflow
```

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
### sentences
sent=[  'the glass of milk',
     'the glass of juice',
     'the cup of tea',
    'I am a good boy',
     'I am a good developer',
     'understand the meaning of words',
     'your videos are good',]
```

```python
sent
```

```python
## Define the vocabulary size
voc_size=10000
```

```python
sent
```

```python
for words in sent:
  print(words)
```

```python
### One Hot Representation
one_hot_repr=[one_hot(words,voc_size) for words in sent]
one_hot_repr
```

```python
## word Embedding Representation

from tensorflow.keras.layers import Embedding
#from tensorflow.keras.processing.sequence import pad_sequences
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
#seq_len    = 100        # fixed padded/truncated sequence length
sent_length=8
embedded_docs=pad_sequences(one_hot_repr,padding='pre',maxlen=sent_length)
print(embedded_docs)
```

```python
## feature representation
embed_dim=10 # embedding vector size
```

```python


from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Input, Embedding

model = Sequential([
    keras.Input(shape=(sent_length,), dtype="int32"),          # explicit Input first
    Embedding(input_dim=voc_size, output_dim=embed_dim, mask_zero=True)  # no input_length here
])

model.compile(optimizer="adam", loss="mse")   # but consider adding a pooling/dense head
```

```python
model.summary()
```

```python
model.predict(embedded_docs)
```
