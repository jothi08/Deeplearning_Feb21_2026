```python
#8. Given a list of numbers, find the sum of all even numbers
def sum_even(num):
  sum=0
  if not num:
    return 0

  for n in num:
    if isinstance(n,(int,float)):
      if n%2==0:
        sum+=n
    else:
      print(f"ignore non numeric charachter:{n}")
  return sum
```

```python
#Note: Slicing creates a new list; it doesn’t modify the original list in place.
my_list = ["apple", "banana", "cherry", "date", "elderberry"]

# Get elements from index 1 up to (but not including) index 3
my_slice = my_list[1:3]  # my_slice will be ["banana", "cherry"]
print(my_slice)
# Get elements from the beginning up to (but not including) index 2
my_slice = my_list[:2]   # my_slice will be ["apple", "banana"]
print(my_slice)

# Get elements from index 2 to the end of the list
my_slice = my_list[2:]   # my_slice will be ["cherry", "date", "elderberry"]
print(my_slice)

# Get every other element (step of 2)
my_slice = my_list[::2]  # my_slice will be ["apple", "cherry", "elderberry"]
print(my_slice)

# Reverse the list (step of -1)
my_slice = my_list[::-1] # my_slice will be ["elderberry", "date", "cherry", "banana", "apple"]
print(my_slice)

# Create a shallow copy of the entire list
my_slice = my_list[:] # my_slice will be a new list identical to my_list
print(my_slice)

# Get the last element
my_slice = my_list[-1:] # my_slice will be ["elderberry"]
print(my_slice)

# Get the last two elements
my_slice = my_list[-2:] # my_slice will be ["date", "elderberry"]
print(my_slice)
```

```python
#10. How to remove all occurrences of a specific element from a list
def remove_all_occurrences(list1,element_to_remove):
  new_list=[]
  for num in list1:
    if num!=element_to_remove:
      new_list.append(num)
  return new_list
```

```python
my_list = [1, 2, 3, 2, 4, 2, 5]
element = 2
new_list = remove_all_occurrences(my_list, element)
print(f"Original list: {my_list}")
print(f"New list (2s removed): {new_list}")
```

```python
'''
17. Given two lists, write a function to find the common elements
Level: Intermediate
Python Code 1: set intersection to finds the common elements between two lists.
'''
list1 = [10, 20, 30, 40]
list2 = [20, 40, 60, 80]

# Convert lists to sets for efficient intersection
set1 = set(list1)
set2 = set(list2)

# Convert back to a list
common_elements = list(set1.intersection(set2))
print(f"Common elements: {common_elements}")
```

```python
#Python Code 2: list comprehension to finds the common elements between two lists
list1 = [10, 20, 30, 40]
list2 = [20, 40, 60, 80]
common_elements = [element for element in list1 if element in list2]
print(f"Common elements: {common_elements}")
```

```python
'''
Nested lists are lists that contain other lists as their elements. Think of it like a list within a list, or even lists within lists within lists – creating a multi-dimensional structure. They’re useful for representing things like matrices, tables, or hierarchical data.
Here’s a simple example:
Python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
This matrix is a nested list. It has three elements, and each of those elements is itself a list.
To access elements in a nested list, we use indexing multiple times. The first index accesses the outer list, and subsequent indices access the inner lists.
For example:
•	matrix[0] gives you [1, 2, 3] (the first inner list).
•	matrix[0][0] gives you 1 (the first element of the first inner list).
•	matrix[1][2] gives you 6 (the third element of the second inner list).
•	matrix[2][1] gives you 8 (the second element of the third inner list).
'''
```

```python
#19. Write a function to flatten a nested list
def flatten_nested_list(nested_list):

    flat_list = []
    for item in nested_list:
        if isinstance(item, list):  # Check if the item is a list
            flat_list.extend(flatten_nested_list(item))  # Recursively flatten sublists
        else:
            flat_list.append(item)  # Add non-list items directly
    return flat_list

# Example Usage:
nested_list1 = [1, 2, [3, 4], 7, 8]
flat_list1 = flatten_nested_list(nested_list1)
print(f"Flattened list: {flat_list1}")
```

```python
def find_longest_string(string_list):

    if not string_list:  # Handle empty list case
        return ""

    longest_string = string_list[0]  # Initialize with the first string
    for string in string_list:
        if len(string) > len(longest_string):
            longest_string = string
    return longest_string

# Example usage:
strings1 = ["apple", "banana", "kiwi", "grapefruit"]
longest1 = find_longest_string(strings1)
print(f"Longest string is: {longest1}")
```

```python
#shallow copy -> If we make changes in new_list, changes will be get applied in original_list too
original_list = [1, 2, [3, 4]]  # Nested list
new_list = original_list[:]  # Shallow copy

new_list[2][0] = 5  # Modify the nested list in the copy

print(original_list)  # Output: [1, 2, [5, 4]]  # Original is also changed!
print(new_list)      # Output: [1, 2, [5, 4]]
```

```python
import copy
##Deep copy -> If we make changes in new_list, changes will not get applied in original_list.
original_list = [1, 2, [3, 4]]
new_list = copy.deepcopy(original_list)  # Deep copy

new_list[2][0] = 5  # Modify the nested list in the copy

print(original_list)  # Output: [1, 2, [3, 4]]  # Original is unchanged!
print(new_list)      # Output: [1, 2, [5, 4]]
```

```python
'''
sort() method (In-place sorting)
The sort() method sorts the list in place, meaning it modifies the original list directly. It doesn’t return a new sorted list;
it changes the order of the elements within the existing list.
Example:
'''
my_list = [5, 1, 8, 2, 9]
my_list.sort()  # Sorts in ascending order by default
print(my_list)  # Output: [1, 2, 5, 8, 9]

my_list.sort(reverse=True)  # Sorts in descending order
print(my_list)  # Output: [9, 8, 5, 2, 1]
```

```python
'''
The sorted() function creates a new sorted list, leaving the original list unchanged. It returns the new sorted list.
Example:
Use sort() when you want to modify the original list directly (in-place sorting).
•	Use sorted() when you want to create a new sorted list without changing the original.
•	Use the key argument to customize the sorting logic based on a specific attribute or transformation of the elements.
'''
my_list = [5, 1, 8, 2, 9]
new_list = sorted(my_list)  # Creates a new sorted list
print(my_list)  # Output: [5, 1, 8, 2, 9] (Original list is unchanged)
print(new_list)  # Output: [1, 2, 5, 8, 9]
```

```python
'''
4. itertools.chain() (For combining multiple iterables):
The itertools.chain() function is useful when you need to combine multiple iterables (not just lists) or
when you have a large number of iterables to combine. It creates an iterator that yields elements from all the input iterables sequentially.
Example:
'''
import itertools

list1 = [1, 2, 3]
list2 = [4, 5, 6]
list3 = [7, 8, 9]

combined_iter = itertools.chain(list1, list2, list3)
combined_list = list(combined_iter) # Convert iterator to list

print(combined_list)  # Output: [1, 2, 3, 4, 5, 6, 7, 8, 9]
```
