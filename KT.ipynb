```python
'''
Q1. Reverse words in a sentence (without split() multiple passes)
Prompt: Given s = "  hello   world  from  chennai  ", return "chennai from world hello".
Key points: strip, multiple spaces, O(n), in-place or two-pass.
'''
```

```python
s = "  hello   world  from  chennai  "
clean_sentence=' '.join(s.split())
print("clean_sentence:",clean_sentence)
```

```python
clean_sentence2=[]
for i in clean_sentence.split(' '):
  clean_sentence2.append(i)
print("clean_sentence2:",clean_sentence2)
```

```python
#clean_sentence2[::-1] is Python list slicing, and this pattern means reverse the list.

print(clean_sentence2[::-1])
#print(clean_sentence2.reverse())

reversed_list = list(reversed(clean_sentence2))
print(reversed_list)
```

```python
'''
Q2. Flatten a nested list (iterative, not recursion depth issues)
Prompt: [1,[2,[3,4],5],6] → [1,2,3,4,5,6].
'''
matrix= [1,[2,[3,4],5],6]
flatten_list=[]
for row in matrix:
  print("row:",row)
  if isinstance(row,list):
    flatten_list.extend(row)
    print("list inside if condition:",flatten_list)
  else:
    flatten_list.append(row)
    print("list inside else condition:",flatten_list)
print("flatten_list:",flatten_list)

```

```python
def flatten(lst):
    out = []
    for x in lst:
        if isinstance(x, list):
            out.extend(flatten(x))
        else:
            out.append(x)
    return out

print(flatten([1, [2, [3, 4], 5], 6]))  # [1, 2, 3, 4, 5, 6]
```
