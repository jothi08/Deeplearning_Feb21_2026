```python
# Step 1: Import Libraries and Load the Model
import numpy as np
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing import sequence
from tensorflow.keras.models import load_model
```

```python
#Set Methods
```

```python
'''
1. Add elements

add(x) → Add one item
update(iterable) → Add multiple items


2. Remove elements

remove(x) → Removes x (error if x not present)
discard(x) → Removes x (no error if missing)
pop() → Removes a random item
clear() → Removes all items


3. Return NEW sets (non-modifying)

union(other) → All items (OR)
intersection(other) → Common items (AND)
difference(other) → Items in A but not B
symmetric_difference(other) → Items not common


4. Modify the current set (in‑place versions)

intersection_update()
difference_update()
symmetric_difference_update()
update() (same as union update)


5. Relationship tests (return True/False)

issubset(other) → A ⊆ B
issuperset(other) → A ⊇ B
isdisjoint(other) → No common items


🎯 Super‑Quick Memory Trick
Add: add, update
Remove: remove, discard, pop, clear
Ops: union, intersection, difference, symmetric_difference
Ops update: same names + _update
Check: issubset, issuperset, isdisjoint
'''
```

```python
#1) Add elements
```

```python
#add(x) — add one item
s = {1, 2}
s.add(3)
print(s) #output -> {1, 2, 3}
```

```python
#update(iterable) — add many items
s = {1, 2}
s.update([2, 3, 4])   # duplicates ignored
print(s)   #output -> {1, 2, 3, 4}
```

```python
#2) Remove elements
```

```python
#remove(x) — error if missing
s = {1, 2, 3}
s.remove(2)
print(s)    #output ->{1, 3}
# s.remove(5)  # KeyError if uncommented
```

```python
#remove(x) — error if missing
s = {1, 2, 3}
s.remove(2)
print(s)    #output ->{1, 3}
s.remove(5)  # KeyError if uncommented
```

```python
#discard(x) — no error if missing
s = {1, 2, 3}
s.discard(2)
print(s)
s.discard(5)  # no error
print(s)
```

```python
#pop() — removes & returns an arbitrary element
s = {10, 20, 30}
x = s.pop()
print("popped:", x)
print("remaining:", s)
x = s.pop()
print("popped:", x)
print("remaining:", s)
```

```python
#clear() — remove all
s = {1, 2, 3}
s.clear()
print(s)
```

```python
#3) Operations that return NEW sets (non‑modifying)
```

```python
#union or |
a = {1, 2, 3}
b = {3, 4, 5}
print(a.union(b))
print(a | b)
```

```python
#intersection or &
print(a.intersection(b))
print(a & b)
```

```python
#difference or -
print(a.difference(b))  # in a, not in b
print(a - b)
```

```python
#symmetric_difference or ^
print(a.symmetric_difference(b))  # in a or b, but not both
print(a ^ b)
```

```python
#Note: All four above do not modify a or b.
```

```python
#4) In‑place (mutating) versions
```

```python
#intersection_update(*others) — keep only common
a = {1, 2, 3}
a.intersection_update({2, 3, 5})
print(a)
#difference_update(*others) — remove elements found in others
a = {1, 2, 3, 4}
a.difference_update({2, 4})
print(a)
#symmetric_difference_update(other) — keep items not common
a = {1, 2, 3}
a.symmetric_difference_update({3, 4})
print(a)
#update(*others) — union update (add all)
a = {1, 2}
a.update({2, 3}, [3, 4])
print(a)
```

```python
#5) Relationship tests (booleans)
```

```python
a = {1, 2}
b = {1, 2, 3}
c = {4, 5}
```

```python
#issubset(other) — A ⊆ B
print(a.issubset(b))
#issuperset(other) — B ⊇ A
print(b.issuperset(a))
#isdisjoint(other) — no common elements
print(a.isdisjoint(c))
```

```python
#Tiny Practice Block (all-in-one)
A = {1, 2, 3}
B = {3, 4}

print("A|B =", A | B)                # {1, 2, 3, 4}
print("A&B =", A & B)                # {3}
print("A-B =", A - B)                # {1, 2}
print("A^B =", A ^ B)                # {1, 2, 4}

A2 = A.copy()
A2.update(B); print("update ->", A2) # {1, 2, 3, 4}

A3 = {1, 2, 3}; A3.intersection_update(B); print("∩= ->", A3)  # {3}
A4 = {1, 2, 3}; A4.difference_update(B);   print("-= ->", A4)  # {1, 2}
A5 = {1, 2, 3}; A5.symmetric_difference_update(B); print("^= ->", A5) # {1, 2, 4}
```
