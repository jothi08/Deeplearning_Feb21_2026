```python
'''
1️⃣ Class and Object (Basics)
✔️ Class: Blueprint for creating objects
✔️ Object: Instance of a class
👉 self means the object that is calling the function.
🚗 Example to Understand Easily
Imagine a class is like a car blueprint, and objects are actual cars.
If you build two cars:
•	Car 1 → Red, Tata
•	Car 2 → White, Mahindra
Each car should remember its own color and brand.
To store each car’s personal data, Python uses self.
📝 In one line:
➡️ self is used to access variables and methods that belong to the object.
'''
```

```python
class Car:
    # Constructor method to initialize instance attributes
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    # Method to describe the car
    def describe(self):
        return f"{self.year} {self.make} {self.model}"

# Creating an object (instance) of the Car class
my_car = Car("Toyota", "Corolla", 2020)

# Accessing attributes and methods
print(my_car.describe())  # Output: 2020 Toyota Corolla
```

```python
'''
The __init__ method in Python is a special method, also known as the constructor,
that’s automatically called when you create a new object from a class
For example, if we have a Dog class, the __init__ method would be where we set the dog’s name, breed,
and age when we create a new dog object.
It ensures that the object has the necessary data to start with.
'''
```

```python
class Dog:
    def __init__(self, name, breed, age):  # Constructor
        self.name = name  # Initialize attributes
        self.breed = breed
        self.age = age

    def bark(self):
        print("Woof!")

# Creating Dog objects:
my_dog = Dog("Buddy", "Golden Retriever", 5)  # __init__ is called automatically
another_dog = Dog("Max", "German Shepherd", 3) # __init__ is called automatically

# Accessing the initialized attributes:
print(my_dog.name)      # Output: Buddy
print(another_dog.age)   # Output: 3
my_dog.bark()         # Output: Woof!
another_dog.bark()  # Output: Woof!
```

```python
'''
Class variables:

class variables are shared among all instances (objects) of a class. They are defined within the class but outside of any method.
Think of them as attributes that belong to the class itself, rather than to individual objects. If you modify a class variable,
that change is reflected for all objects of that class. It can be accessed using either the class name or an instance.
'''
```

```python
'''
Class Methods:

Class methods are bound to the class and not the instance of the class. They take the class itself (cls) as the first argument
instead of the instance (self). You define them using the @classmethod decorator.
'''
```

```python
'''
species is a class variable, shared by all Dog instances.
name is an instance variable, unique to each Dog.
describe_species is a class method, accessed using Dog.describe_species().
It can access the class variable species using cls.species.
'''
class Dog:
    species = "Canine"  # Class variable

    def __init__(self, name):
        self.name = name  # Instance variable

    @classmethod
    def describe_species(cls):  # Class method
        print(f"All dogs are {cls.species}.")

dog1 = Dog("Buddy")
dog2 = Dog("Max")

print(dog1.name)       # Output: Buddy
print(dog2.name)       # Output: Max

print(Dog.species)    # Output: Canine (accessing class variable)

Dog.describe_species() # Output: All dogs are Canine. (calling class method)
```

```python
'''
Key Concepts
1. Class Variable (A.x)
•	Shared by all instances unless overridden by instance.
•	Initially: A.x = 1 🔍
2. Instance Variable (self.y, also a1.x when assigned)
•	Specific to each object.
________________________________________
🔄 Execution Explanation
Step 1:
a1 = A() → object a1 created
a2 = A() → object a2 created
Both objects see A.x = 1
Step 2:
a1.x = 10
This does NOT modify the class variable.
It creates a new instance variable x inside a1.
So now:
•	a1.x → 10 (instance variable overrides class variable)
•	a2.x → still comes from class variable → 1
Step 3:
A.x = 5
Updates the class variable.
So:
•	a1.x → still 10 (instance variable)
•	a2.x → now 5 (updated class variable)
•	A.x → 5
________________________________________
🟢 Final Output
Python
10 5 5
'''
```

```python
class A:
    x = 1              # Class variable

    def __init__(self):
        self.y = 2     # Instance variable

a1 = A()
a2 = A()
a1.x = 10             # Creates an instance variable x for a1
A.x = 5               # Updates the class variable x

print(a1.x, a2.x, A.x) #output -> 10 5 5
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Generic animal sound")

    def eat(self):
        print("Eating...")
#Now, let’s say we want to create a Dog class. A dog is a animal, so we can use inheritance:

class Dog(Animal):  # Dog inherits from Animal
    def bark(self):
        print("Woof!")

    def speak(self): # Overriding the speak method
        print("Woof!")
'''
Dog automatically gets the name attribute and the speak() and eat() methods from Animal. We don’t have to redefine them.
Dog can also have its own methods, like bark(), which are specific to dogs.
Dog can override methods from the parent class. For example, the speak() method in Dog overrides the speak() method in Animal
to provide a more specific “Woof!” sound. This is polymorphism in action.
Now, we can create Dog objects:
'''
my_dog = Dog("Buddy")
print(my_dog.name)  # Output: Buddy (inherited from Animal)
my_dog.speak()      # Output: Woof! (overridden method)
my_dog.eat()        # Output: Eating... (inherited method)
my_dog.bark()      # Output: Woof! (dog-specific method)
'''
Inheritance helps us avoid writing the same code multiple times. We define common attributes and methods in the base class,
and then more specialized classes inherit those features, adding or modifying what’s necessary.
It makes our code more organized, reusable, and easier to maintain.
It’s important to remember that inheritance models an “is-a” relationship. A Dog is a Animal. “
'''
```

```python
#method overriding -> subclass overrides the method of parent class
class Animal:
    def speak(self):
        print("Generic animal sound")
#And then we have a Dog class that inherits from Animal:

class Dog(Animal):
    def speak(self):  # Overriding the speak() method
        print("Woof!")
#The Dog class overrides the speak() method. Now, if we create a Dog object and call speak():

my_dog = Dog()
my_dog.speak()  # Output: Woof!
```

```python
'''
What’s happening?
•	A, B, and C each define a show() method.
•	B and C both override A.show().
•	D inherits from both B and C (class D(B, C)), and doesn’t define its own show().
When you call d.show(), Python looks for the method using the Method Resolution Order (MRO) for D.
For D(B, C), the MRO is:
D → B → C → A → object
So Python finds show() in B first and uses that.
'''
class A:
    def show(self):
        print("A")
class B(A):
    def show(self):
        print("B")
class C(A):
    def show(self):
        print("C")

class D(B,C):
    pass
d=D()
d.show()  #-> output B
```

```python
'''
what is public and private variable , write program to determine their behaviour
✅ Public vs Private Variables in Python
🔹 Public Variables
•	Can be accessed anywhere: inside the class, outside the class, by any object.
•	No special notation.
•	Default visibility in Python.
Example:
Python
self.x = 10
Show more lines
________________________________________
🔹 Private Variables
•	Declared using a double underscore prefix: __variable
•	Python performs name mangling:
__x becomes _ClassName__x
•	Cannot be accessed directly outside the class.
•	Used for internal data protection.
Example:
Python
self.__y = 20
'''
```

```python
'''
Public (x)
•	Accessible everywhere.
Protected (_x)
•	Accessible inside class and subclasses.
•	Can be accessed outside class but considered bad practice.
Private (__x)
•	Name-mangled → cannot be accessed directly from outside.
•	Safest form of encapsulation.
Type	Syntax	Intended Access
Public	x	Anywhere
Protected	_x	Within class and subclasses
Private	__x	Only inside class (name-mangled)
'''
```

```python
class Demo:
  def __init__(self):
    self.public = 10 # Public variable
    self.__private = 20 # Private variable

  def show(self):
    print("Public:", self.public)
    print("Private:", self.__private)


obj = Demo()

# Accessing public variable
print("Accessing public directly:", obj.public)

# Accessing private variable directly (will cause error)
try:
  print(obj.__private)
except AttributeError:
  print("Cannot access private variable directly!")

# Accessing private variable using name mangling
print("Accessing private using name mangling:", obj._Demo__private)

# Calling method that prints both
obj.show()
```
