```python
'''
Class variables:

class variables are shared among all instances (objects) of a class. They are defined within the class but outside of any method.
Think of them as attributes that belong to the class itself, rather than to individual objects. If you modify a class variable,
that change is reflected for all objects of that class. It can be accessed using either the class name or an instance.
'''
```

```python
'''
Class Methods:

Class methods are bound to the class and not the instance of the class. They take the class itself (cls) as the first argument
instead of the instance (self). You define them using the @classmethod decorator.
'''
```

```python
'''
species is a class variable, shared by all Dog instances.
name is an instance variable, unique to each Dog.
describe_species is a class method, accessed using Dog.describe_species().
It can access the class variable species using cls.species.
'''
class Dog:
    species = "Canine"  # Class variable

    def __init__(self, name):
        self.name = name  # Instance variable

    @classmethod
    def describe_species(cls):  # Class method
        print(f"All dogs are {cls.species}.")

dog1 = Dog("Buddy")
dog2 = Dog("Max")

print(dog1.name)       # Output: Buddy
print(dog2.name)       # Output: Max

print(Dog.species)    # Output: Canine (accessing class variable)

Dog.describe_species() # Output: All dogs are Canine. (calling class method)
```

```python
'''
Key Concepts
1. Class Variable (A.x)
•	Shared by all instances unless overridden by instance.
•	Initially: A.x = 1 🔍
2. Instance Variable (self.y, also a1.x when assigned)
•	Specific to each object.
________________________________________
🔄 Execution Explanation
Step 1:
a1 = A() → object a1 created
a2 = A() → object a2 created
Both objects see A.x = 1
Step 2:
a1.x = 10
This does NOT modify the class variable.
It creates a new instance variable x inside a1.
So now:
•	a1.x → 10 (instance variable overrides class variable)
•	a2.x → still comes from class variable → 1
Step 3:
A.x = 5
Updates the class variable.
So:
•	a1.x → still 10 (instance variable)
•	a2.x → now 5 (updated class variable)
•	A.x → 5
________________________________________
🟢 Final Output
Python
10 5 5
'''
```

```python
class A:
    x = 1              # Class variable

    def __init__(self):
        self.y = 2     # Instance variable

a1 = A()
a2 = A()
a1.x = 10             # Creates an instance variable x for a1
A.x = 5               # Updates the class variable x

print(a1.x, a2.x, A.x) #output -> 10 5 5
```

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print("Generic animal sound")

    def eat(self):
        print("Eating...")
#Now, let’s say we want to create a Dog class. A dog is a animal, so we can use inheritance:

class Dog(Animal):  # Dog inherits from Animal
    def bark(self):
        print("Woof!")

    def speak(self): # Overriding the speak method
        print("Woof!")
'''
Dog automatically gets the name attribute and the speak() and eat() methods from Animal. We don’t have to redefine them.
Dog can also have its own methods, like bark(), which are specific to dogs.
Dog can override methods from the parent class. For example, the speak() method in Dog overrides the speak() method in Animal
to provide a more specific “Woof!” sound. This is polymorphism in action.
Now, we can create Dog objects:
'''
my_dog = Dog("Buddy")
print(my_dog.name)  # Output: Buddy (inherited from Animal)
my_dog.speak()      # Output: Woof! (overridden method)
my_dog.eat()        # Output: Eating... (inherited method)
my_dog.bark()      # Output: Woof! (dog-specific method)
'''
Inheritance helps us avoid writing the same code multiple times. We define common attributes and methods in the base class,
and then more specialized classes inherit those features, adding or modifying what’s necessary.
It makes our code more organized, reusable, and easier to maintain.
It’s important to remember that inheritance models an “is-a” relationship. A Dog is a Animal. “
'''
```

```python
#method overriding -> subclass overrides the method of parent class
class Animal:
    def speak(self):
        print("Generic animal sound")
#And then we have a Dog class that inherits from Animal:

class Dog(Animal):
    def speak(self):  # Overriding the speak() method
        print("Woof!")
#The Dog class overrides the speak() method. Now, if we create a Dog object and call speak():

my_dog = Dog()
my_dog.speak()  # Output: Woof!
```

```python
'''
What’s happening?
•	A, B, and C each define a show() method.
•	B and C both override A.show().
•	D inherits from both B and C (class D(B, C)), and doesn’t define its own show().
When you call d.show(), Python looks for the method using the Method Resolution Order (MRO) for D.
For D(B, C), the MRO is:
D → B → C → A → object
So Python finds show() in B first and uses that.
'''
class A:
    def show(self):
        print("A")
class B(A):
    def show(self):
        print("B")
class C(A):
    def show(self):
        print("C")

class D(B,C):
    pass
d=D()
d.show()  #-> output B
```

```python
'''
2️⃣ Encapsulation
Encapsulation means restricting direct access to variables and methods.
Encapsulation means hiding data inside a class and allowing access only through methods.
✔️ Public
✔️ Protected (_var)
✔️ Private (__var)
'''
```

```python
class Parent:
  def __init__(self):
    self.public = "Public Variable"
    self._protected = "Protected Variable"
    self.__private = "Private Variable"

class Child(Parent):
  def access_variables(self):
    print("Inside Child class:")
    print("Public:", self.public)
    print("Protected:", self._protected)


# Create objects
p = Parent()
c = Child()

print("Access from object:")
print("Public:", p.public)
print("Protected:", p._protected) # Allowed but not recommended
# Private access fails
try:
  print("Private:", p.__private)
except AttributeError:
  print("Private: Cannot access private variable")

# Accessing private variable using name mangling
print("Accessing private using name mangling by parent:", p._Parent__private)
print("Accessing private using name mangling by child:", c._Parent__private)

print("\nAccess from subclass:")
c.access_variables()
```

```python
class Demo:
  def __init__(self):
    self.public = 10 # Public variable
    self.__private = 20 # Private variable

  def show(self):
    print("Public:", self.public)
    print("Private:", self.__private)


obj = Demo()

# Accessing public variable
print("Accessing public directly:", obj.public)

# Accessing private variable directly (will cause error)
try:
  print(obj.__private)
except AttributeError:
  print("Cannot access private variable directly!")

# Accessing private variable using name mangling
print("Accessing private using name mangling:", obj._Demo__private)

# Calling method that prints both
obj.show()
```

```python
'''
👉 Why this is encapsulation?

__name and __marks are private → cannot be accessed directly.
Access is only through getters and setters → controlled and safe.
'''
class Student:
    def __init__(self, name, marks):
        self.__name = name      # private variable
        self.__marks = marks    # private variable

    # getter method
    def get_marks(self):
        return self.__marks

    # setter method
    def set_marks(self, marks):
        if marks < 0 or marks > 100:
            print("Invalid marks")
        else:
            self.__marks = marks

# usage
s = Student("Jothi", 85)

print(s.get_marks())   # access using method → 85
s.set_marks(95)        # modify safely
print(s.get_marks())   # 95

# s.__marks → not allowed (private)
```

```python
'''
Abstraction (ABC)
✅ What is Abstraction?
Abstraction means showing only the important things and hiding the unnecessary details from the user.
It lets you work with an object without knowing how it works internally.
'''
```

```python
'''
👉 Why this is abstraction?

Vehicle is an abstract class → you cannot create its object.
It contains an abstract method start() → no details provided.
Car gives the actual implementation.
The user only calls start() — they don’t need to know:

how the engine works
how fuel is burned
how ignition happens
Only important information is shown.
'''
from abc import ABC, abstractmethod

# abstract class
class Vehicle(ABC):

    @abstractmethod
    def start(self):
        pass  # hiding the internal details

# child class
class Car(Vehicle):
    def start(self):
        print("Car engine starts with a key")

# usage
v = Car()
v.start()   # user only calls start(), internal mechanism hidden
```

```python
'''
✅ What is Polymorphism?
Polymorphism means one function name doing different things depending on the object.
In simple words:

Same method
Different behavior Based on the object that calls it
'''
```

```python
'''
👉 Why this is polymorphism?

Both Dog and Cat have a method named sound().
The function make_sound() calls sound(), but:

For Dog → it prints "Dog barks"
For Cat → it prints "Cat meows"

Same function name → different result → polymorphism.
'''
class Dog:
    def sound(self):
        print("Dog barks")

class Cat:
    def sound(self):
        print("Cat meows")

# function showing polymorphism
def make_sound(animal):
    animal.sound()   # same method name, different behavior

# usage
d = Dog()
c = Cat()

make_sound(d)   # Dog barks
make_sound(c)   # Cat meows
```

```python
'''
✅ What is Method Overriding in polymorphism?
Method overriding happens when a child class provides a new version of a method that already exists in the parent class.

Same method name
Same parameters
Different behavior
Defined in child class
'''
```

```python
'''
👉 Why this is method overriding?

The parent class Animal has a method sound().
The child class Dog redefines the same method.
When d.sound() is called, Python uses the child class method, not the parent one.

This is the core idea of runtime polymorphism.
'''
class Animal:
    def sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    # overriding parent's sound() method
    def sound(self):
        print("Dog barks")

# usage
a = Animal()
d = Dog()

a.sound()   # Animal makes a sound
d.sound()   # Dog barks (overridden method)
```

```python
'''
✅ What is Operator Overloading (dunder methods)?
Operator overloading lets you use operators like +, -, *, ==, etc. with your own class objects.
Example:

Using + to add two numbers → normal
Using + to add two objects (like Points) → operator overloading

Python allows this using special methods like __add__(), __mul__(), __str__(), etc.
'''
```

```python
'''
👉 Simple Explanation

p1 + p2 doesn't work by default for objects.
But when we define __add__(), Python understands how to add them.
So p1 + p2 becomes:
p1.add(p2)
That returns a new Point object with added coordinates.
'''
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    # operator overloading for +
    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def __str__(self):
        return f"({self.x}, {self.y})"

# usage
p1 = Point(2, 3)
p2 = Point(4, 5)

p3 = p1 + p2       # calls __add__()
print(p3)          # (6, 8)
```

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return Number(self.value + other.value)

    def __str__(self):
        return str(self.value)

print(Number(10) + Number(5))  # prints 15
```
