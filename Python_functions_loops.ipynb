```python
'''
3. What are *args and kwargs in Python functions?
The *args and kwargs you to write flexible and reusable functions that can accept an arbitrary number of arguments.
You can use both in a function, but *args must come before **kwargs in the function definition.

*args
*args allows a function to accept any number of positional arguments.
Inside the function, args is treated as a tuple containing all the additional positional arguments passed to the function.


**kwargs
**kwargs allows a function to accept any number of keyword arguments.
Inside the function, kwargs is treated as a dictionary where the keys are the argument names and
the values are the corresponding values passed to the function.

Practical use case:
Functions that need to handle a flexible number of arguments.
Wrapper functions or decorators where the exact arguments are not known in advance.
By using *args and **kwargs, you can write more general, reusable, and adaptable code.
'''
```

```python
def example_function(*args, **kwargs):
    print("Args:", args)
    print("Kwargs:", kwargs)

example_function(1, 2, 3, name="Alice", age=25)
# Output:
# Args: (1, 2, 3)
# Kwargs: {'name': 'Alice', 'age': 25}
```

```python
'''
9. What are decorators in Python? How are they used?
A decorator in Python is a function that modifies another function’s behavior without changing its code, using the @ syntax.”

Use Cases:
Logging: Automatically log function calls or outputs.
Access Control: Enforce permissions or user roles.
Caching: Store results of expensive function calls.
Timing: Measure execution time of functions.
Decorators are a key feature of Python for creating reusable, modular, and cleaner code.
'''
```

```python
def my_decorator(func):
    def wrapper():
        print("Before function runs")
        func()
        print("After function runs")
    return wrapper

@my_decorator
def say_hello():
    print("Hello!")

say_hello()
```

```python
'''
11. Can functions return multiple values? If yes, then how?

Yes, functions in Python can return multiple values by using tuples, lists, dictionaries, or other data structures.

A function can return multiple values by separating them with commas. These values are implicitly packed into a tuple,
which can then be unpacked when the function is called.
'''
```

```python
def calculate(a, b):
    return a + b, a - b, a * b

# Calling the function
result = calculate(10, 5)
print(result)  # Output: (15, 5, 50)

# Unpacking the result
add, sub, mul = calculate(10, 5)
print(add)  # Output: 15
print(sub)  # Output: 5
print(mul)  # Output: 50
```

```python
#12. What is a default, positional, and keyword argument in a function?
'''
1. Default Arguments
Default arguments allow you to specify a default value for a parameter in the function definition.
If no value is provided during the function call, the default value is used.
'''
def greet(name, message="Hello"):
    print(f"{message}, {name}!")

greet("Alice")          # Output: Hello, Alice!
greet("Bob", "Hi")      # Output: Hi, Bob!
```

```python
'''
13. What is a recursive function and how do you create it?
Level: Intermediate

A recursive function in Python is a function that calls itself during its execution.
This allows the function to solve problems by breaking them into smaller sub-problems of the same type.
'''
```

```python
def factorial(n):
    # Base case
    if n == 0 or n == 1:
        return 1
    # Recursive case
    return n * factorial(n - 1)

# Test the recursive function
print(factorial(5))  # Output: 120
```

```python
'''
14. Can we assign a new name to a function and call it through the new name? If yes, then how?
Level: Intermediate

Yes, in Python, you can assign a new name to an existing function and call it using the new name.
Functions in Python are treated as first-class objects, meaning they can be assigned to variables,
passed as arguments, or returned from other functions.

When you assign a new name to a function, it doesn’t create a copy of the function.
The new name is just another reference to the same function object.
'''
```

```python
def greet(name):
    print(f"Hello, {name}!")

# Assign a new name to the function
welcome = greet

# Call the function using the new name
welcome("Alice")  # Output: Hello, Alice!

# Original function name still works
greet("Bob")  # Output: Hello, Bob!
```

```python
'''
15. How to create an inner function?
Level: Intermediate

An inner function is a function defined inside another function. It is also known as a nested function.
Inner functions are typically used when a function is only relevant to the enclosing (outer) function or needs access to its variables.

Example:
'''
def outer_function(name):
    def inner_function():
        print(f"Hello, {name}!")  # Accessing outer function's variable

    # Call the inner function
    inner_function()

# Call the outer function
outer_function("Alice")  # Output: Hello, Alice!
#Encapsulation: Inner functions are hidden from the outside and can only be accessed within the outer function, keeping the code modular.
```

```python
# Lambda function
add = lambda a, b: a + b
print(add(3, 5))  # Output: 8
```

```python
# Double each number using map()
numbers = [1, 2, 3, 4, 5]
doubled = map(lambda x: x * 2, numbers)
print(list(doubled))  # Output: [2, 4, 6, 8, 10]
```

```python
# Filter even numbers
evens = filter(lambda x: x % 2 == 0, numbers)
print(list(evens))  # Output: [2, 4]
```

```python
#Sorting with Custom Keys:

words = ["apple", "banana", "cherry"]
sorted_words = sorted(words, key=lambda x: len(x))
print(sorted_words)  # Output: ['apple', 'banana', 'cherry']
```

```python
from functools import reduce

numbers = [1, 2, 3, 4, 5]

# Calculate the product of all numbers
product = reduce(lambda x, y: x * y, numbers)
print(product)  # Output: 120
```

```python
#19. What is the global keyword in a function?
#The global keyword is used to modify a global variable inside a function.

x = 10
def modify_global():
    global x
    x += 5

modify_global()
print(x)  # Output: 15
```

```python
'''
✅ What is a nonlocal variable in Python?
A nonlocal variable is a variable defined in an outer (enclosing) function, which can be modified inside an inner function using the nonlocal keyword.
It tells Python:
👉 “This variable is not local to the inner function — use the one from the outer function.”
'''
def outer():
    x = 10   # outer variable

    def inner():
        nonlocal x   # refers to outer function's x
        x = 20
        print("Inner x:", x)

    inner()
    print("Outer x:", x)

outer()
```
