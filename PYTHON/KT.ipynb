```python
'''
Q1. Reverse words in a sentence (without split() multiple passes)
Prompt: Given s = "  hello   world  from  chennai  ", return "chennai from world hello".
Key points: strip, multiple spaces, O(n), in-place or two-pass.
'''
```

```python
s = "  hello   world  from  chennai  "
clean_sentence=' '.join(s.split())
print("clean_sentence:",clean_sentence)
```

```python
clean_sentence2=[]
for i in clean_sentence.split(' '):
  clean_sentence2.append(i)
print("clean_sentence2:",clean_sentence2)
```

```python
#clean_sentence2[::-1] is Python list slicing, and this pattern means reverse the list.
'''
list[start : stop : step]
start → where to begin
stop → where to end
step → how to move (positive = forward, negative = backward)
clean_sentence2[::-1]
start is empty → start at the end
stop is empty → go until the beginning
step = -1 → move backwards one step at a time

Therefore, it produces a reversed copy of the list.
[::-1] does not modify the original list.
It creates a new reversed list.
'''
print(clean_sentence2[::-1])
#print(clean_sentence2.reverse())

reversed_list = list(reversed(clean_sentence2))
print(reversed_list)
```

```python
'''
Q2. Flatten a nested list (iterative, not recursion depth issues)
Prompt: [1,[2,[3,4],5],6] → [1,2,3,4,5,6].
'''
matrix= [1,[2,[3,4],5],6]
flatten_list=[]
for row in matrix:
  print("row:",row)
  if isinstance(row,list):
    flatten_list.extend(row)
    print("list inside if condition:",flatten_list)
  else:
    flatten_list.append(row)
    print("list inside else condition:",flatten_list)
print("flatten_list:",flatten_list)

```

```python
def flatten(lst):
    out = []
    for x in lst:
        if isinstance(x, list):
            out.extend(flatten(x))
        else:
            out.append(x)
    return out

print(flatten([1, [2, [3, 4], 5], 6]))  # [1, 2, 3, 4, 5, 6]
```

```python
def is_palindrome(s: str) -> bool:
    t = ''.join(ch.lower() for ch in s if ch.isalnum())
    return t == t[::-1]
```

```python
#Is palindrome (ignore cases & non-alphanumerics)
#Keeps only letters and digits (isalnum()).
#Example: "A man, a plan, a canal: Panama" → ['a','m','a','n','a','p','l','a','n','a','c','a','n','a','l','p','a','n','a','m','a'].
'''
Compares mirrored characters from the start (i) and end (j).
If any mismatch → not a palindrome.
If the loop finishes → it is a palindrome.
'''
def is_palindrome(s: str) -> bool:
    t = [ch.lower() for ch in s if ch.isalnum()]
    i, j = 0, len(t) - 1
    while i < j:
        if t[i] != t[j]:
            return False
        i += 1; j -= 1
    return True
```

```python
#3) Remove duplicates from a list (keep order)
def remove_dups_keep_order(nums):
    seen, out = set(), []
    for x in nums:
        if x not in seen:
            seen.add(x); out.append(x)
    return out
```

```python
'''
🔍 Binary Search (Super Simple Explanation)
Imagine you have a sorted list of numbers, like:
[2, 4, 6, 8, 10, 12, 14]

You want to find 10.
Instead of checking one-by-one:

Look at the middle number.
If it is your number → stop, found!
If your number is smaller, search the left side.
If your number is bigger, search the right side.
Repeat until you find it.

You keep cutting the list in half, so it becomes very fast.

'''
```

```python
def binary_search(arr, target):
    left = 0
    right = len(arr) - 1

    while left <= right:
        mid = (left + right) // 2   # middle index

        if arr[mid] == target:
            return mid               # found!
        elif target < arr[mid]:
            right = mid - 1         # search left half
        else:
            left = mid + 1          # search right half

    return -1  # not found
```

```python
arr = [2, 4, 6, 8, 10, 12, 14]
print(binary_search(arr, 10))
```
